```python
import random

# Smart home grid
# S = Start, H = Home/Goal, X = Obstacle
grid = [
    ["S", ".", "."],
    ["X", ".", "."],
    [".", ".", "H"]
]

# Starting position
state = (0, 0)

# Possible actions
actions = ["Up", "Down", "Left", "Right"]

# Function to move the robot
def move(state, action):
    row, col = state

    if action == "Up":
        row -= 1
    elif action == "Down":
        row += 1
    elif action == "Left":
        col -= 1
    elif action == "Right":
        col += 1

    # Check boundaries
    if row < 0 or row >= 3 or col < 0 or col >= 3:
        return state, -1

    # Check obstacle
    if grid[row][col] == "X":
        return state, -5

    # Goal reached
    if grid[row][col] == "H":
        return (row, col), 10

    # Normal movement
    return (row, col), -1


# Robot navigation
total_reward = 0
state = (0, 0)

print("Robot starting at:", state)

for step in range(10):

    if state == (2, 2):
        break

    action = random.choice(actions)
    next_state, reward = move(state, action)

    print("\nStep:", step + 1)
    print("State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)

    total_reward += reward
    state = next_state

print("\nFinal State:", state)
print("Total Reward:", total_reward)

if state == (2, 2):
    print("Robot reached home successfully!")
else:
    print("Robot did not reach home.")
```
