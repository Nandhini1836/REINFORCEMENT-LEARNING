```python
import random

# States
states = [
    "Start",
    "Good Position",
    "Opponent Weak",
    "Winning",
    "Losing"
]

# Actions
actions = [
    "Move Pawn",
    "Move Knight",
    "Attack"
]

# Rewards
rewards = {
    "Winning": 10,
    "Losing": -10,
    "Good Position": 2,
    "Opponent Weak": 5,
    "Start": 0
}

# Transition probabilities
transitions = {
    "Start": {
        "Move Pawn": ["Good Position", "Opponent Weak"],
        "Move Knight": ["Good Position", "Losing"],
        "Attack": ["Opponent Weak", "Losing"]
    },
    
    "Good Position": {
        "Move Pawn": ["Opponent Weak", "Winning"],
        "Move Knight": ["Good Position", "Losing"],
        "Attack": ["Winning", "Losing"]
    },
    
    "Opponent Weak": {
        "Move Pawn": ["Winning", "Opponent Weak"],
        "Move Knight": ["Winning", "Losing"],
        "Attack": ["Winning", "Losing"]
    }
}

# Select next state
def get_next_state(state, action):
    if state in transitions and action in transitions[state]:
        return random.choice(transitions[state][action])
    return state

# Simulate a game
state = "Start"
total_reward = 0

print("Initial State:", state)

for step in range(5):
    if state in ["Winning", "Losing"]:
        break

    action = random.choice(actions)
    next_state = get_next_state(state, action)
    reward = rewards[next_state]

    print("\nStep:", step + 1)
    print("State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)

    total_reward += reward
    state = next_state

print("\nFinal State:", state)
print("Total Reward:", total_reward)
```
