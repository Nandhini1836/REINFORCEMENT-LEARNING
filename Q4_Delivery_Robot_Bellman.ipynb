```python
# Delivery robot using Bellman Equation

# Travel costs between locations
cost = {
    "A": {"B": 2, "C": 5},
    "B": {"C": 1, "D": 4},
    "C": {"D": 2},
    "D": {}
}

# Bellman values
V = {
    "A": float("inf"),
    "B": float("inf"),
    "C": float("inf"),
    "D": 0
}

# Calculate minimum cost
for _ in range(3):
    for state in ["C", "B", "A"]:
        if state in cost and cost[state]:
            V[state] = min(
                cost[state][next_state] + V[next_state]
                for next_state in cost[state]
            )

print("Minimum travel cost from each location:")

for state in V:
    print(state, ":", V[state])

# Find optimal path
state = "A"
path = [state]

while state != "D":
    next_state = min(
        cost[state],
        key=lambda x: cost[state][x] + V[x]
    )

    path.append(next_state)
    state = next_state

print("\nOptimal Path:", " -> ".join(path))
print("Minimum Travel Cost:", V["A"])
```
