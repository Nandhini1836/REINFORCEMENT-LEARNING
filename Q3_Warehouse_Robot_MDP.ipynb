```python
import random

# States
states = ["Start", "Aisle", "Shelf", "Charging", "Goal"]

# Actions
actions = ["Move", "Pick Item", "Charge"]

# Rewards
rewards = {
    "Start": 0,
    "Aisle": -1,
    "Shelf": 5,
    "Charging": -2,
    "Goal": 10
}

# Transition probabilities
transitions = {
    "Start": {
        "Move": ["Aisle", "Aisle"],
        "Pick Item": ["Start"],
        "Charge": ["Charging"]
    },

    "Aisle": {
        "Move": ["Shelf", "Charging"],
        "Pick Item": ["Aisle"],
        "Charge": ["Charging"]
    },

    "Shelf": {
        "Move": ["Goal", "Aisle"],
        "Pick Item": ["Goal", "Shelf"],
        "Charge": ["Charging"]
    }
}

# Function to select next state
def get_next_state(state, action):
    if state in transitions and action in transitions[state]:
        return random.choice(transitions[state][action])
    return state

# Simulation
state = "Start"
total_reward = 0

print("Initial State:", state)

for step in range(5):

    if state == "Goal":
        break

    action = random.choice(actions)
    next_state = get_next_state(state, action)
    reward = rewards[next_state]

    print("\nStep:", step + 1)
    print("State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)

    total_reward += reward
    state = next_state

print("\nFinal State:", state)
print("Total Reward:", total_reward)
```
