```python
import random

# Advertisement options
ads = ["Ad 1", "Ad 2", "Ad 3"]

# True engagement probabilities
engagement_rate = [0.3, 0.6, 0.4]

# Number of times each ad is selected
count = [0, 0, 0]

# Total rewards for each ad
total_reward = [0, 0, 0]

# Exploration rate
epsilon = 0.1

# Number of users
trials = 100

for i in range(trials):

    # Exploration or exploitation
    if random.random() < epsilon:
        # Exploration
        choice = random.randint(0, 2)
    else:
        # Exploitation
        average_reward = [
            total_reward[j] / count[j] if count[j] > 0 else 0
            for j in range(3)
        ]
        choice = average_reward.index(max(average_reward))

    # Simulate user engagement
    reward = 1 if random.random() < engagement_rate[choice] else 0

    count[choice] += 1
    total_reward[choice] += reward

# Calculate average rewards
average_reward = [
    total_reward[i] / count[i] if count[i] > 0 else 0
    for i in range(3)
]

print("Advertisement Results:\n")

for i in range(3):
    print(ads[i])
    print("Times selected:", count[i])
    print("Total engagement:", total_reward[i])
    print("Average reward:", round(average_reward[i], 2))
    print()

best_ad = average_reward.index(max(average_reward))

print("Best Advertisement:", ads[best_ad])
print("Maximum Engagement Rate:", round(average_reward[best_ad], 2))
```
